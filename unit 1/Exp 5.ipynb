```python
!pip install transformers torch

from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

prompt = "Artificial Intelligence is transforming the world by"

result = generator(
    prompt,
    max_new_tokens=50,
    num_return_sequences=1,
    do_sample=True,
    temperature=0.7,
    pad_token_id=50256
)

generated_text = result[0]["generated_text"]

print("Input Prompt:")
print(prompt)

print("\nGenerated Response:")
print(generated_text)

print("\nResponse Analysis:")
print("Total Characters:", len(generated_text))
print("Total Words:", len(generated_text.split()))
print("Prompt Words:", len(prompt.split()))
print("New Words Generated:", len(generated_text.split()) - len(prompt.split()))
print("Unique Words:", len(set(generated_text.lower().split())))
print("Vocabulary Diversity:", round(
    len(set(generated_text.lower().split())) /
    max(len(generated_text.split()), 1), 2
))
```
