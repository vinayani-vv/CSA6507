```python
!pip install transformers

from transformers import BertTokenizer, GPT2Tokenizer

text = "Artificial Intelligence is amazing!"

bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
gpt_tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

bert_tokens = bert_tokenizer.tokenize(text)
gpt_tokens = gpt_tokenizer.tokenize(text)

print("Original Text:", text)

print("\nBERT Tokenizer")
print("Tokens:", bert_tokens)
print("Token IDs:", bert_tokenizer.convert_tokens_to_ids(bert_tokens))
print("Number of Tokens:", len(bert_tokens))

print("\nGPT Tokenizer")
print("Tokens:", gpt_tokens)
print("Token IDs:", gpt_tokenizer.convert_tokens_to_ids(gpt_tokens))
print("Number of Tokens:", len(gpt_tokens))

print("\nComparison")
print("BERT Token Count:", len(bert_tokens))
print("GPT Token Count:", len(gpt_tokens))
```
